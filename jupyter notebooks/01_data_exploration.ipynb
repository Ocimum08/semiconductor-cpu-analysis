```python
# Select individual product records only
cpu_products = df[
    (df["Entity kind"] == "sku") &
    (
        df["Object name"].str.contains(
            r"\b(Ryzen|Core i[3579]|Xeon|EPYC|Pentium|Celeron|"
            r"Threadripper|Athlon)\b",
            case=False,
            na=False,
            regex=True
        )
    )
].copy()

print("Potential CPU product specification records:", len(cpu_products))
print("Unique CPU product IDs:", cpu_products["Object ID"].nunique())
print("Manufacturers:", cpu_products["Company"].nunique())

display(
    cpu_products[
        ["Object ID", "Object name", "Company", "Specification", "Value", "Unit"]
    ].head(15)
)
```

```python
# Find the most common specifications among selected products
display(
    cpu_products["Specification"]
    .value_counts()
    .head(25)
    .to_frame("Record_Count")
)
```

```python
clock_rows = cpu_products[
    cpu_products["Specification"].str.contains(
        r"clock|frequency",
        case=False,
        na=False
    )
]

display(
    clock_rows[
        ["Object name", "Company", "Specification", "Value", "Unit"]
    ].head(30)
)

print("\nClock units:")
display(clock_rows["Unit"].value_counts(dropna=False).to_frame("Count"))
```

```python
# 1. Reshape the data: one row per CPU product
product_table = (
    cpu_products.pivot_table(
        index=["Object ID", "Object name", "Company"],
        columns="Specification",
        values="Value",
        aggfunc="first"
    )
    .reset_index()
)

product_table.columns.name = None

# 2. Convert selected specification fields to numeric values
numeric_fields = [
    "Cores", "Enabled CPU cores", "Threads",
    "L2 cache", "L3 cache",
    "Listed TDP", "Default TDP",
    "Record year", "Release year"
]

for col in numeric_fields:
    if col in product_table.columns:
        product_table[col] = pd.to_numeric(
            product_table[col], errors="coerce"
        )

# 3. Standardize core counts and TDP field names
if "Cores" in product_table.columns and "Enabled CPU cores" in product_table.columns:
    product_table["CPU cores"] = (
        product_table["Cores"].combine_first(
            product_table["Enabled CPU cores"]
        )
    )
elif "Cores" in product_table.columns:
    product_table["CPU cores"] = product_table["Cores"]
elif "Enabled CPU cores" in product_table.columns:
    product_table["CPU cores"] = product_table["Enabled CPU cores"]

if "Listed TDP" in product_table.columns and "Default TDP" in product_table.columns:
    product_table["TDP (W)"] = (
        product_table["Listed TDP"].combine_first(
            product_table["Default TDP"]
        )
    )
elif "Listed TDP" in product_table.columns:
    product_table["TDP (W)"] = product_table["Listed TDP"]
elif "Default TDP" in product_table.columns:
    product_table["TDP (W)"] = product_table["Default TDP"]

# 4. Standardize base and boost clocks to GHz
clock_long = cpu_products[
    cpu_products["Specification"].isin(["Base clock", "Boost clock"])
].copy()

clock_long["Value_num"] = pd.to_numeric(
    clock_long["Value"], errors="coerce"
)

clock_long["Unit"] = clock_long["Unit"].str.strip()

clock_long["Clock_GHz"] = np.where(
    clock_long["Unit"].eq("MHz"),
    clock_long["Value_num"] / 1000,
    np.where(
        clock_long["Unit"].eq("GHz"),
        clock_long["Value_num"],
        np.nan
    )
)

clock_wide = clock_long.pivot_table(
    index="Object ID",
    columns="Specification",
    values="Clock_GHz",
    aggfunc="first"
).reset_index()

clock_wide = clock_wide.rename(columns={
    "Base clock": "Base clock (GHz)",
    "Boost clock": "Boost clock (GHz)"
})

# Remove raw clock columns to avoid confusing them with normalized values
product_table = product_table.drop(
    columns=["Base clock", "Boost clock"],
    errors="ignore"
)

product_table = product_table.merge(
    clock_wide,
    on="Object ID",
    how="left",
    validate="one_to_one"
)

print("Product-level table shape:", product_table.shape)
print("Unique product IDs:", product_table["Object ID"].nunique())

display(
    product_table[
        [
            "Object name", "Company", "CPU cores", "Threads",
            "Base clock (GHz)", "Boost clock (GHz)", "TDP (W)"
        ]
    ].head(10)
)
```

```python
company_summary = (
    product_table.groupby("Company")
    .agg(
        Products=("Object ID", "nunique"),
        Products_with_core_data=("CPU cores", "count"),
        Products_with_base_clock=("Base clock (GHz)", "count"),
        Products_with_boost_clock=("Boost clock (GHz)", "count"),
        Products_with_TDP=("TDP (W)", "count"),
        Median_base_clock_GHz=("Base clock (GHz)", "median"),
        Median_boost_clock_GHz=("Boost clock (GHz)", "median"),
        Median_TDP_W=("TDP (W)", "median")
    )
    .reset_index()
)

display(company_summary.round(2))
```

```python
comparison = (
    product_table.groupby("Company")
    .agg(
        Products=("Object ID", "nunique"),
        Core_data_available=("CPU cores", "count"),
        Threads_available=("Threads", "count"),
        Base_clock_available=("Base clock (GHz)", "count"),
        Boost_clock_available=("Boost clock (GHz)", "count"),
        TDP_available=("TDP (W)", "count")
    )
    .reset_index()
)

for metric in [
    "Core_data_available",
    "Threads_available",
    "Base_clock_available",
    "Boost_clock_available",
    "TDP_available"
]:
    comparison[metric + "_pct"] = (
        comparison[metric] / comparison["Products"] * 100
    ).round(1)

display(comparison)
```
